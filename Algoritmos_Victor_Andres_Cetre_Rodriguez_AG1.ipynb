##Torres de Hanoi con Divide y vencerás

```python
#Torres de Hanoi - Divide y venceras
#T(n) = 1 + 2*T(n-1)  ->  O(2^n)
def Torres_Hanoi(N, desde, hasta):
  if N == 1:
    print("Lleva la ficha ", desde, " hasta ", hasta)
  else:
    Torres_Hanoi(N-1, desde, 6-desde-hasta)
    print("Lleva la ficha ", desde, " hasta ", hasta)
    Torres_Hanoi(N-1, 6-desde-hasta, hasta)


Torres_Hanoi(3, 1, 3)
```

##Devolución de cambio por técnica voraz

```python
def cambio_monedas(N, SM):
  SOLUCION = [0]*len(SM)
  ValorAcumulado = 0

  for i, valor in enumerate(SM):
    monedas = (N-ValorAcumulado)//valor
    SOLUCION[i] = monedas
    ValorAcumulado = ValorAcumulado + monedas*valor

    if ValorAcumulado == N:
      return SOLUCION

  return None  #No se completo la cantidad exacta con este sistema


print(cambio_monedas(15, [25, 10, 5, 1]))

#La técnica voraz no siempre es óptima: con un sistema no canónico [4,3,1]
#para la cantidad 6 devuelve 4+1+1 (3 monedas) en vez del óptimo 3+3 (2 monedas)
print(cambio_monedas(6, [4, 3, 1]))
```

##N-Reinas por técnica de vuelta atrás

```python
def escribe(S):
  n = len(S)
  for x in range(n):
    print("")
    for i in range(n):
      if S[i] == x+1:
        print(" X ", end="")
      else:
        print(" - ", end="")
  print("")


def es_prometedora(SOLUCION, etapa):
  for i in range(etapa+1):
    if SOLUCION.count(SOLUCION[i]) > 1:
      return False
    for j in range(i+1, etapa+1):
      if abs(i-j) == abs(SOLUCION[i]-SOLUCION[j]):
        return False
  return True


#solucion=None en vez de [] para evitar el problema del argumento por defecto mutable
def reinas(N, solucion=None, etapa=0, mostrar_tablero=False):
  if solucion is None:
    solucion = [0 for _ in range(N)]

  for i in range(1, N+1):
    solucion[etapa] = i

    if es_prometedora(solucion, etapa):
      if etapa == N-1:
        print(solucion)
        if mostrar_tablero:
          escribe(solucion)
          print()
      else:
        reinas(N, solucion, etapa+1, mostrar_tablero)

    solucion[etapa] = 0


reinas(8)
```

```python
reinas(4, mostrar_tablero=True)
```

##Viaje por el rio. Programación dinámica

```python
TARIFAS = [
[0,   5,   4,   3,   999, 999, 999],
[999, 0,   999, 2,   3,   999, 11],
[999, 999, 0,   1,   999, 4,   10],
[999, 999, 999, 0,   5,   6,   9],
[999, 999, 999, 999, 0,   999, 4],
[999, 999, 999, 999, 999, 0,   3],
[999, 999, 999, 999, 999, 999, 0]
]


#P(i,j) = min { T(i,j) , P(i,k)+T(k,j) para todo i<k<=j }
def Precios(TARIFAS):
  N = len(TARIFAS[0])

  PRECIOS = [[9999]*N for _ in range(N)]
  RUTA = [[""]*N for _ in range(N)]

  for i in range(0, N-1):
    RUTA[i][i] = i
    PRECIOS[i][i] = 0
    for j in range(i+1, N):
      MIN = TARIFAS[i][j]
      RUTA[i][j] = i

      for k in range(i, j):
        if PRECIOS[i][k] + TARIFAS[k][j] < MIN:
          MIN = min(MIN, PRECIOS[i][k] + TARIFAS[k][j])
          RUTA[i][j] = k
        PRECIOS[i][j] = MIN

  return PRECIOS, RUTA


PRECIOS, RUTA = Precios(TARIFAS)

print("PRECIOS")
for fila in PRECIOS:
  print(fila)

print("\nRUTA")
for fila in RUTA:
  print(fila)


#Ruta reconstruida con recursividad a partir de la matriz RUTA
def calcular_ruta(RUTA, desde, hasta):
  if desde == hasta:
    return [desde]
  return calcular_ruta(RUTA, desde, RUTA[desde][hasta]) + [hasta]


print("\nRuta de 0 a 6:", calcular_ruta(RUTA, 0, 6))
print("Coste total:", PRECIOS[0][6])
```

##Práctica individual: los dos puntos más cercanos

```python
#Fuerza bruta 1D - O(n^2)
def cercanos_fuerza_bruta_1d(puntos):
  n = len(puntos)
  mejor_par = None
  mejor_dist = float("inf")

  for i in range(n):
    for j in range(i+1, n):
      d = abs(puntos[i] - puntos[j])
      if d < mejor_dist:
        mejor_dist = d
        mejor_par = (puntos[i], puntos[j])

  return mejor_par, mejor_dist


ejemplo_1d = [3403, 4537, 9089, 9746, 7259, 100, 250, 4600]
print(cercanos_fuerza_bruta_1d(ejemplo_1d))
```

```python
#Divide y venceras 1D - O(n log n). Solo se comparan los puntos cerca de la frontera
def cercanos_divide_venceras_1d(puntos):

  def resuelve(arr):
    n = len(arr)
    if n <= 3:
      return cercanos_fuerza_bruta_1d(arr)

    mitad = n // 2
    valor_medio = arr[mitad]

    mejor_izq = resuelve(arr[:mitad])
    mejor_der = resuelve(arr[mitad:])
    mejor = mejor_izq if mejor_izq[1] <= mejor_der[1] else mejor_der

    d = mejor[1]
    candidatos_izq = [p for p in arr[:mitad] if valor_medio - p < d]
    candidatos_der = [p for p in arr[mitad:] if p - valor_medio < d]

    for pi in candidatos_izq:
      for pd in candidatos_der:
        if abs(pi - pd) < mejor[1]:
          mejor = ((pi, pd), abs(pi - pd))

    return mejor

  return resuelve(sorted(puntos))


print(cercanos_divide_venceras_1d(ejemplo_1d))
```

```python
#Extension a 2D. En la combinacion solo se mira la "franja" cercana a la linea divisoria
import math


def distancia_2d(p, q):
  return math.hypot(p[0]-q[0], p[1]-q[1])


def cercanos_fuerza_bruta_2d(puntos):
  n = len(puntos)
  mejor_par = None
  mejor_dist = float("inf")
  for i in range(n):
    for j in range(i+1, n):
      d = distancia_2d(puntos[i], puntos[j])
      if d < mejor_dist:
        mejor_dist = d
        mejor_par = (puntos[i], puntos[j])
  return mejor_par, mejor_dist


def cercanos_divide_venceras_2d(puntos):
  puntos_x = sorted(puntos, key=lambda p: p[0])

  def resuelve(arr):
    n = len(arr)
    if n <= 3:
      return cercanos_fuerza_bruta_2d(arr)

    mitad = n // 2
    punto_medio = arr[mitad]

    mejor_izq = resuelve(arr[:mitad])
    mejor_der = resuelve(arr[mitad:])
    mejor = mejor_izq if mejor_izq[1] <= mejor_der[1] else mejor_der

    d = mejor[1]
    franja = [p for p in arr if abs(p[0]-punto_medio[0]) < d]
    franja.sort(key=lambda p: p[1])

    n_franja = len(franja)
    for i in range(n_franja):
      for j in range(i+1, min(i+7, n_franja)):
        dd = distancia_2d(franja[i], franja[j])
        if dd < mejor[1]:
          mejor = ((franja[i], franja[j]), dd)

    return mejor

  return resuelve(puntos_x)


import random
random.seed(42)
puntos_2d = [(random.uniform(0, 1000), random.uniform(0, 1000)) for _ in range(30)]

print(cercanos_fuerza_bruta_2d(puntos_2d))
print(cercanos_divide_venceras_2d(puntos_2d))
```

```python
#Extension a 3D
def distancia_3d(p, q):
  return math.sqrt(sum((a-b)**2 for a, b in zip(p, q)))


def cercanos_fuerza_bruta_3d(puntos):
  n = len(puntos)
  mejor_par = None
  mejor_dist = float("inf")
  for i in range(n):
    for j in range(i+1, n):
      d = distancia_3d(puntos[i], puntos[j])
      if d < mejor_dist:
        mejor_dist = d
        mejor_par = (puntos[i], puntos[j])
  return mejor_par, mejor_dist


def cercanos_divide_venceras_3d(puntos):
  puntos_x = sorted(puntos, key=lambda p: p[0])

  def resuelve(arr):
    n = len(arr)
    if n <= 3:
      return cercanos_fuerza_bruta_3d(arr)

    mitad = n // 2
    punto_medio = arr[mitad]

    mejor_izq = resuelve(arr[:mitad])
    mejor_der = resuelve(arr[mitad:])
    mejor = mejor_izq if mejor_izq[1] <= mejor_der[1] else mejor_der

    d = mejor[1]
    franja = [p for p in arr if abs(p[0]-punto_medio[0]) < d]

    n_franja = len(franja)
    for i in range(n_franja):
      for j in range(i+1, n_franja):
        if franja[j][0] - franja[i][0] >= d:
          break
        dd = distancia_3d(franja[i], franja[j])
        if dd < mejor[1]:
          mejor = ((franja[i], franja[j]), dd)

    return mejor

  return resuelve(puntos_x)


puntos_3d = [(random.uniform(0, 1000), random.uniform(0, 1000), random.uniform(0, 1000)) for _ in range(30)]

print(cercanos_fuerza_bruta_3d(puntos_3d))
print(cercanos_divide_venceras_3d(puntos_3d))
```

```python
#Comprobacion de la mejora de complejidad: fuerza bruta O(n^2) vs divide y venceras O(n log n)
import time

for n in [200, 400, 800, 1600]:
  muestra = [(random.uniform(0, 100000), random.uniform(0, 100000)) for _ in range(n)]

  t0 = time.perf_counter()
  cercanos_fuerza_bruta_2d(muestra)
  t1 = time.perf_counter()

  cercanos_divide_venceras_2d(muestra)
  t2 = time.perf_counter()

  print("n=", n, " fuerza_bruta=", round(t1-t0, 4), "s  divide_venceras=", round(t2-t1, 4), "s")
```
